##Tarea 3 · Las cámaras de frío, con variables
Cada tienda tiene 3 cámaras de frío, y cada cámara envía su temperatura 12 veces por hora, las 24 horas del día.

```python
tiendas = 120
camaras_por_tienda = 3
lecturas_por_hora = 12
lecturas_al_dia = (lecturas_por_hora * 24)
lecturas_al_anio = (lecturas_al_dia * 365)
print(tiendas * camaras_por_tienda * lecturas_al_dia)
print(tiendas * camaras_por_tienda * lecturas_al_anio)
```

##Tarea 4 · De lecturas a megas
Cada lectura ocupa unos 46 bytes. ¿Cuánto ocupan las lecturas de un día? ¿Y las de un año?

```python
bytes_por_lectura = 46
megas_al_dia = (tiendas * camaras_por_tienda * lecturas_al_dia * bytes_por_lectura / 1000000)
gigas_al_anio = (tiendas * camaras_por_tienda * lecturas_al_anio * bytes_por_lectura / 1000000000)
print(f"Sensores: {megas_al_dia:.2f}  MB al día")
print(f"Sensores: {gigas_al_anio:.2f}  GB al año")
```

##Tarea 5 · Los clics de la web y la app
La web y la app de Colmena Market reciben unas 250.000 visitas al día. En cada visita se registran unos 12 clics, y cada clic ocupa unos 220 bytes.

Calcula, con variables, cuántos clics hay al día, cuántos megas ocupan y cuántos gigas son al año.

```python
visitas_web_dia = 250000
clics_por_visita = 12
bytes_por_clic = 220

clics_al_dia = (clics_por_visita * visitas_web_dia)
bytes_al_dia = (clics_al_dia * 220)
megas_al_dia = (bytes_al_dia / 1000000)
gigas_al_anio = (bytes_al_dia * 365 / 1000000000)

print(f"Clics: {clics_al_dia} al día")
print(f"Clics: {megas_al_dia:.2f} MB al día")
print(f"Clics: {gigas_al_anio:.2f} GB al año")
```
